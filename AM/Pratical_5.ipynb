#Karl Pearson's Coeffient of corelation

```python
import numpy as np

# Sample data
x = np.array([5, 20, 30, 40, 50])
y = np.array([30, 50, 55, 60, 70])

# Calculate Pearson correlation coefficient
mean_x = np.mean(x)
mean_y = np.mean(y)

numerator = np.sum((x - mean_x) * (y - mean_y))
denominator = np.sqrt(np.sum((x - mean_x) ** 2) * np.sum((y - mean_y) ** 2))
r = numerator / denominator

print("Karl Pearson's Coefficient of Correlation (r):", round(r, 2))
```

#Spearman's Rank Correlation

```python
import numpy as np
from scipy.stats import rankdata

# Sample data
x = np.array([10, 20, 30, 40, 50])
y = np.array([10, 25, 30, 45, 55])

# Rank the data
rank_x = rankdata(x)
rank_y = rankdata(y)

# Calculate differences and Spearman's rho
d = rank_x - rank_y
n = len(x)
rho = 1 - (6 * np.sum(d ** 2)) / (n * (n ** 2-1))

print("Spearman's Rank Correlation (p):", round(rho, 4))
```

#Simple Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
x = np.array([10, 20, 30, 40, 50])
y = np.array([25, 45, 65, 75, 95])

# Step 1: Calculate means
x_mean = np.mean(x)
y_mean = np.mean(y)

# Step 2: Calculate slope (b) and intercept (a)
numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean) ** 2)
b = numerator / denominator
a = y_mean - b * x_mean

print(f"Slope (b) = {b:.2f}")
print(f"Intercept (a) = {a:.2f}")

# Step 3: Predicted values
y_pred = a + b * x
print(y_pred)
# Step 4: Plot
plt.scatter(x, y, color='blue', label='Actual data')
plt.plot(x, y_pred, color='red', label='Regression Line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Simple Linear Regression (Manual Calculation)')
plt.legend()
plt.grid(True)
plt.show()
```

#Multiple Linear Regression

```python
import numpy as np

# Dataset
X1 = np.array([230, 44, 17, 151, 180])
X2 = np.array([37, 39, 45, 41, 10])
Y = np.array([22, 10, 9, 18, 13])

# Form X matrix with a column of 1s for intercept
X = np.column_stack((np.ones(len(X1)), X1, X2))

#Convert Y to column vector
Y = Y.reshape(-1, 1)

# Apply Normal Equation: b = (X^T X)^(-1) X^T Y
b = np.linalg.inv(X.T @ X) @ X.T @ Y

#Extract coefficients
a, b1, b2 = b.flatten()


print(f"Intercept (a): {a:.3f}")
print(f"Coefficient for X1 (b1): {b1:.3f}")
print(f"Coefficient for X2 (b2): {b2:.3f}")
print(f"Regression Equation: Y = {a:.3f} + {b1:.3f}*X1 + {b2:.3f}*X2")

# Predict for a new example
TV_new = 100
Radio_new = 30
Y_pred = a + b1 * TV_new + b2 * Radio_new


print(f"\nPredicted Sales for TV={TV_new}, Radio={Radio_new} = {Y_pred:.2f}")
```
