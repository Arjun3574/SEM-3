#1.Perform Large Sample Test for Single Mean (Z-test)

```python
# ****** Import Required Libraries ******
import numpy as np
from scipy.stats import norm

# ****** Z-Test Function Definition ******
def z_test(data_points, hypothesized_mean, std=None, significance_level=0.05, test_type='two-sided'):
    data_points = np.array(data_points)
    sample_size = len(data_points)
    sample_mean = np.mean(data_points)

    if std is None:
        std = np.std(data_points, ddof=1)

    # Calculate Z-score using formula: (x̄ - μ) / (σ / √n)
    z_score = (sample_mean - hypothesized_mean) / (std / np.sqrt(sample_size))

    # Calculate p-value Based on Test Type
    if test_type == 'two-sided':
        p_val = 2 * (1 - norm.cdf(abs(z_score)))
    elif test_type == 'greater':
        p_val = 1 - norm.cdf(z_score)
    elif test_type == 'less':
        p_val = norm.cdf(z_score)
    else:
        raise ValueError("test_type must be 'two-sided', 'less', or 'greater'")

    # Determine whether to reject the null hypothesis
    reject_null = p_val < significance_level

    # Return results in a dictionary
    return {
        "Sample Mean": sample_mean,
        "Hypothesized Mean": hypothesized_mean,
        "Z-Score": z_score,
        "p-value": p_val,
        "Conclusion": "Reject H0" if reject_null else "Fail to Reject H0"
    }

# Sample Data and Hypothesis
dataset = [75, 56, 84, 25, 99, 88, 84, 83, 75, 64, 48, 45]
mu_hypothesis = 50
alpha_val = 0.05

# ****** Run Z-Test and Display Results ******
test_result = z_test(dataset, mu_hypothesis, std=None, significance_level=alpha_val, test_type='two-sided')
print("Sample Mean:", test_result["Sample Mean"])
print("Hypothesized Mean:", test_result["Hypothesized Mean"])
print("Z-Score:", test_result["Z-Score"])
print("p-value:", test_result["p-value"])
print("Conclusion:", test_result["Conclusion"])
```

#2 .Perform Small Samplet-test for Difference of Means

```python

from scipy import stats

print("=== Small Sample t-Test for Difference of Means ===")

# -----------------------------------------------
# Example: Marks scored by two groups of students
# -----------------------------------------------
group_A = [82, 88, 90, 85, 87]   # Sample 1
group_B = [78, 75, 80, 74, 77]   # Sample 2

# Step 1: Perform independent t-test using SciPy
t_stat, p_value = stats.ttest_ind(group_A, group_B)

# Step 2: Display calculated values
print(f"\nGroup A Marks: {group_A}")
print(f"Group B Marks: {group_B}")
print(f"\nt-statistic = {t_stat:.3f}")
print(f"p-value = {p_value:.3f}")

# Step 3: Decision rule (significance level = 0.05)
alpha = 0.05
if p_value < alpha:
    print("\n Reject H0 → There is a significant difference in the means of the two groups.")
else:
    print("\n Fail to reject H0 → No significant difference between the two group means.")

# -----------------------------------------------
# Step 4: Interpretation
# -----------------------------------------------
print("\n--- INTERPRETATION ---")
print("H₀ : μ₁ = μ₂  (There is no significant difference between the two groups)")
print("H₁ : μ₁ ≠ μ₂  (There is a significant difference between the two groups)")
print("Decision is based on the p-value compared to α = 0.05.")
```

#3.Perform Test for Single Proportion(LargeSample)

```python
# Practical 3 – Hypothesis Testing
# Z-Test for Single Proportion

import math

print("=== Z-Test for Single Proportion ===")

# Example: Out of 1000 voters, 540 favor a candidate
# Claimed population proportion = 0.50

n = 1000
x = 540
p0 = 0.50
p_hat = x / n

Z = (p_hat - p0) / math.sqrt((p0 * (1 - p0)) / n)

print(f"Sample Proportion = {p_hat:.3f}, Z = {Z:.3f}")

if abs(Z) > 1.96:
    print("Reject H0 → Proportion significantly different from 0.5.")
else:
    print("Fail to reject H0 → No significant difference in proportion.")
```

#4.Perform F-test for Comparisonof Variances(SmallSamples)

```python
# Practical 3 – Hypothesis Testing
# F-Test for Comparison of Variances

from scipy import stats

print("=== F-Test for Comparison of Variances ===")

# Example: Compare variance of test scores between two batches
batch_1 = [14, 15, 16, 15, 14, 13]
batch_2 = [10, 12, 11, 13, 12, 11]

var1 = stats.tvar(batch_1)
var2 = stats.tvar(batch_2)

F = var1 / var2 if var1 > var2 else var2 / var1
df1 = len(batch_1) - 1
df2 = len(batch_2) - 1

F_critical = stats.f.ppf(0.95, df1, df2)

print(f"Variance 1 = {var1:.3f}, Variance 2 = {var2:.3f}")
print(f"F = {F:.3f}, Critical F = {F_critical:.3f}")

if F > F_critical:
    print("Reject H0 → Variances are significantly different.")
else:
    print("Fail to reject H0 → No significant difference in variances.")
```
