#Lagrange Interpolation Formula

```python

def lagrange_interpolation(x_values, y_values, x):

    n = len(x_values)
    result = 0.0

    for i in range(n):
        term = y_values[i]
        for j in range(n):
            if j != i:
                term *= (x - x_values[j]) / (x_values[i] - x_values[j])
        result += term

    return result


# Given data points
x_values = [5, 7, 11, 13]
y_values = [150, 392, 1452, 2366]

# Value to estimate
x_to_find = 10

# Apply Lagrange Interpolation
y_estimated = lagrange_interpolation(x_values, y_values, x_to_find)

# (I also fixed the f-string formatting error from your last attempt)
print(f"Estimated value of f({x_to_find}) = {y_estimated:.2f}")
```

#ImplementNewton'sInterpolationmethod

```python
import numpy as np
import math

# Function to calculate divided differences
def forward_diff_table(y):
    n = len(y)
    diff_table = np.zeros((n, n))
    diff_table[:, 0] = y

    for j in range(1, n):
        for i in range(n - j):
            diff_table[i][j] = diff_table[i + 1][j - 1] - diff_table[i][j - 1]

    return diff_table

# Function to evaluate Newton polynomial at x_val
def newton_forward_interpolation(x, y, x_val):
    h = x[1] - x[0] # uniform spacing
    diff_table = forward_diff_table(y)

    u = (x_val - x[0]) / h
    result = y[0]
    u_term = 1

    for i in range(1, len(y)):
        u_term *= (u - (i - 1))
        result += (u_term / math.factorial(i)) * diff_table[0][i]

    return result

# Given data
x = np.array([5, 7, 9, 11])
y = np.array([150, 392, 1452, 2366])

# Value to estimate
x_val = 10

# Apply Newton's Interpolation
y_pred = newton_forward_interpolation(x, y, x_val)


print(f"Estimated value of f({x_val}) = {y_pred:.2f}")
```

#Apply LU Decomposition for Linear Systems

```python
import numpy as np
from scipy.linalg import lu

# --- 1. Define the System Ax = b ---
# This is our system:
#  2x + 1y - 1z = 8
# -3x - 1y + 2z = -11
# -2x + 1y + 2z = -3

A = np.array([
    [ 2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
])

b = np.array([8, -11, -3])

print("Original Matrix A:\n", A)
print("\nOriginal Vector b:\n", b)

# --- 2. Decompose A (PA = LU) ---
# The lu function returns the P, L, and U matrices
P, L, U = lu(A)

print("\nPermutation Matrix P:\n", P)
print("\nLower Triangular L:\n", L)
print("\nUpper Triangular U:\n", U)

# --- 3. Solve the Two-Step System ---

# We are solving PAx = Pb, which is LUx = Pb

# Step A: Solve Ly = Pb for y (Forward Substitution)
# First, calculate Pb
Pb = P @ b
y = np.linalg.solve(L, Pb)

print("\nIntermediate vector y (solution to Ly = Pb):\n", y)

# Step B: Solve Ux = y for x (Backward Substitution)
x = np.linalg.solve(U, y)

print("\n--- Final Solution x --- \n", x)

# --- 4. Verification ---
# Check if the solution is correct by plugging x back into A @ x
solution_check = A @ x
print("\nVerification (A @ x):\n", solution_check)
print("Should be equal to b:\n", b)
```
