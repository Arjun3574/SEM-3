#Perform First-Order Markov Process Simulation

```python
import numpy as np

states = [0, 1]  # 0 = State A, 1 = State B

# Transition Probability Matrix: P[current_state] gives probabilities to go to [A, B]
P = [
    [0.1, 0.9],  # From A to [A, B]
    [0.8, 0.2],  # From B to [A, B]
]

current_state = 0
n = 10  # number of steps

print("Markov Process Simulation:")
sequence = [current_state]

for i in range(n):
    # sample next state using the probability row for the current state
    current_state = np.random.choice(states, p=P[current_state])
    sequence.append(int(current_state))

print("Sequence of States:", sequence)
```

# Calculate n-Step Transition Probability.

```python
import numpy as np

P = np.array([[0.1, 0.9],
              [0.8, 0.2]])

n = 3
Pn = np.linalg.matrix_power(P, n)
print("P^n =\n", Pn)
# probability of going from state 0 to state 1 in n steps:
print("p_0,1^({}) =".format(n), Pn[0,1])
```

#Calculate Steady State Distribution

```python
import numpy as np

# Transition matrix
P = np.array([[0.7, 0.3],
[0.4, 0.6]])

#vP = v , v(P - I) = 0
A = (P - np.eye(2)).T
b = np.zeros(2)

# Add the normalization condition: n1 + n2 = 1
A = np.vstack((A, np.ones(2)))
b = np.append(b, 1)
print(A)
print(b)
# Solve
steady_state = np.linalg.lstsq(A, b, rcond=None)[0]
print("Steady-state distribution:", steady_state)
```

#Simulate Higher-Order Markov Process(Second-Order)

```python
import numpy as np
import random

# Define states
states = [0, 1, 2] # A, B, C

# Define transition probabilities for each pair (previous two states)
P = {
(0, 0): [0.7, 0.2, 0.1],
(0, 1): [0.3, 0.5, 0.2],
(0, 2): [0.6, 0.2, 0.2],
(1, 0): [0.6, 0.3, 0.1],
(1, 1): [0.2, 0.7, 0.1],
(1, 2): [0.3, 0.4, 0.3],
(2, 0): [0.2, 0.5, 0.3],
(2, 1): [0.3, 0.3, 0.4],
(2, 2): [0.5, 0.2, 0.3]
}

# Initialize with two starting states
current_state = [random.choice(states), random.choice(states)]
sequence = current_state.copy()

n = 5 # number of steps

# Simulation loop
for i in range(n):
  prev_pair = (sequence[-2], sequence[-1])
  probs = P[prev_pair]
  next_state = np.random.choice(states, p=probs)
  sequence.append(next_state)

print("Second-Order Markov Process Simulation:")
print("Sequence of States:", sequence)
```
